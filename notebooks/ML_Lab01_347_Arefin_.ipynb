```python
## Pandas Exercises : Task

import pandas as pd
data = {
'student': ['Asha', 'Babu', 'Chayan', 'Dipa', 'Emon', 'Farah'],
'study_hours': [2, 5, 1, 7, 3, 6],
'sleep_hours': [8, 6, 9, 5, 7, 6],
'attendance': [70, 88, 55, 95, 75, 91],
'snacks': [3, 2, 5, 1, 4, 2],
'passed': ['Yes', 'Yes', 'No', 'Yes', 'No', 'Yes']
}
df = pd.DataFrame(data)
display(df)
print("Features are : study_hours,sleep_hours,attendance,snacks,passed")
print("Label is : passed")
print('Number of sample presents : ',df.shape)
print("\n\nStudents list whom attendacne percentage is greater than 80 : \n",df[df['attendance'] >= 80])
print("\n\nAverage study hours : ",df[df['passed'] == 'Yes']['study_hours'].mean())
print("\n\nSorting the Dataframe by attendace in descending order : \n",df.sort_values('attendance', ascending=False))

print('''\n\nNo, this tiny dataset does not prove that snack purchases cause failure.
It only shows a pattern or possible correlation between snacks and passing/failing.
Correlation can be a clue, but it does not prove causation because other factors,
 such as study hours, attendance, or sleep, may also affect the result.''')
```
